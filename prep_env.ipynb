```python
from sklearn.impute import KNNImputer

df['vibration'] = df['vibration'].fillna(df['vibration'].median())
df['temperature_C'] = df['temperature_C'].fillna(df['temperature_C'].median())

imputer = KNNImputer(n_neighbors=5)
cols_knn = ['pression_bar', 'production_jour', 'temperature_C', 'vibration']
df[cols_knn] = imputer.fit_transform(df[cols_knn])
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest

num_cols = ['pression_bar', 'vibration', 'temperature_C', 'production_jour']

outliers_summary = pd.DataFrame(index=num_cols)

def detect_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    return (df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))

outliers_summary['IQR'] = [detect_iqr(df, col).sum() for col in num_cols]

def detect_zscore(df, col):
    z = (df[col] - df[col].mean()) / df[col].std()
    return np.abs(z) > 3

outliers_summary['Z-Score'] = [detect_zscore(df, col).sum() for col in num_cols]

def detect_modified_zscore(df, col):
    median = df[col].median()
    mad = np.median(np.abs(df[col] - median))
    # Précaution si MAD vaut 0
    if mad == 0:
        return pd.Series(False, index=df.index)
    mod_z = 0.6745 * np.abs(df[col] - median) / mad
    return mod_z > 3.5

outliers_summary['Modified Z-Score'] = [detect_modified_zscore(df, col).sum() for col in num_cols]

iso = IsolationForest(contamination=0.05, random_state=42)
df_clean_num = df[num_cols].dropna()
preds = iso.fit_predict(df_clean_num)
outliers_summary['Isolation Forest'] = [0] * len(num_cols)
print("--- Récapitulatif du nombre d'outliers détectés ---")
print(outliers_summary)
print(f"\nIsolation Forest (détection globale multivariée) : {(preds == -1).sum()} lignes suspectes.")

# Visualisation par Boxplots
plt.figure(figsize=(12, 6))
for i, col in enumerate(num_cols, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot - {col}')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest

num_cols = ['pression_bar', 'vibration', 'temperature_C', 'production_jour']

# 1. Calcul des masques booléens pour chaque méthode
# IQR
q1 = df[num_cols].quantile(0.25)
q3 = df[num_cols].quantile(0.75)
iqr = q3 - q1
mask_iqr = ((df[num_cols] < (q1 - 1.5 * iqr)) | (df[num_cols] > (q3 + 1.5 * iqr))).any(axis=1)

# Z-Score (|Z| > 3)
z_scores = (df[num_cols] - df[num_cols].mean()) / df[num_cols].std()
mask_zscore = (np.abs(z_scores) > 3).any(axis=1)

# Modified Z-Score / MAD (> 3.5)
medians = df[num_cols].median()
mads = np.median(np.abs(df[num_cols] - medians), axis=0)
mod_z_scores = 0.6745 * np.abs(df[num_cols] - medians) / mads
mask_mod_zscore = (mod_z_scores > 3.5).any(axis=1)

# Isolation Forest
iso = IsolationForest(contamination=0.05, random_state=42)
# Remplacement temporaire des NaN pour exécuter l'algo multivarié
preds = iso.fit_predict(df[num_cols].fillna(df[num_cols].median()))
mask_iso = (preds == -1)

# 2. Construction du tableau récapitulatif ligne par ligne
df_outliers = df.copy()
df_outliers['IQR'] = mask_iqr
df_outliers['Z_Score'] = mask_zscore
df_outliers['Mod_Z_Score'] = mask_mod_zscore
df_outliers['Isolation_Forest'] = mask_iso

# Filtrer uniquement les lignes détectées par au moins une méthode
mask_global = mask_iqr | mask_zscore | mask_mod_zscore | mask_iso
tableau_valeurs_aberrantes = df_outliers[mask_global]

# Affichage des colonnes clés et des détections
cols_to_show = num_cols + ['IQR', 'Z_Score', 'Mod_Z_Score', 'Isolation_Forest']
tableau_valeurs_aberrantes[cols_to_show]
```
